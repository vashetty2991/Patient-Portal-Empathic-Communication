```python

# Figure 2: LLM Empathic Opportunity Classification Performance
# This notebook generates the publication-ready Figure 2 used in the JAMIA submission.
#
# Notes:
# - No patient-level data or PHI are included.
# - Values are aggregate F1 scores and bootstrap confidence intervals.
# - Figure uses Open Sans when available, with a sans-serif fallback.
# - Error bars are drawn first without caps; filled markers are drawn last for accessibility.

import matplotlib
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib import font_manager
import numpy as np
from pathlib import Path

# ---------- Typography ----------
available_fonts = {f.name for f in font_manager.fontManager.ttflist}
font_name = "Open Sans" if "Open Sans" in available_fonts else "DejaVu Sans"

plt.rcParams.update({
    "font.family": font_name,
    "font.size": 10.5,
    "axes.labelsize": 10.5,
    "xtick.labelsize": 9.5,
    "ytick.labelsize": 9.5,
    "legend.fontsize": 10.5,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "svg.fonttype": "none",
})

# ---------- Colors ----------
# Okabe-Ito colorblind-safe palette
basic_color = "#0072B2"       # blue
codebook_color = "#D55E00"    # vermillion
text_color = "#111111"
grid_color = "#CFCFCF"
spine_color = "#777777"

prompt_labels = ["Baseline", "Conditional", "Anon + Cond."]

def err_array(vals, lows, highs):
    vals = np.asarray(vals, dtype=float)
    lows = np.asarray(lows, dtype=float)
    highs = np.asarray(highs, dtype=float)
    return np.vstack([vals - lows, highs - vals])

# ---------- Aggregate results ----------
# Each entry contains:
# ([F1 values], [lower CI bounds], [upper CI bounds])
panels = {
"Empathic\nOpportunity":{
"8B Zero Basic":([0.8000,0.8348,0.8317],[0.7702,0.8025,0.7993],[0.8295,0.8631,0.8629]),
"8B Zero Code":([0.7343,0.8601,0.8089],[0.7000,0.8328,0.7677],[0.7650,0.8892,0.8383]),
"8B Few Basic":([0.7899,0.8331,0.8538],[0.7597,0.8061,0.8242],[0.8184,0.8605,0.8827]),
"8B Few Code":([0.7587,0.8318,0.8528],[0.7236,0.8016,0.8228],[0.7903,0.8592,0.8820]),
"70B Zero Basic":([0.7970,0.8155,0.7760],[0.7612,0.7825,0.7424],[0.8299,0.8482,0.8125]),
"70B Zero Code":([0.8246,0.8494,0.8474],[0.7936,0.8193,0.8101],[0.8530,0.8794,0.8723]),
"70B Few Basic":([0.8420,0.8489,0.8608],[0.8147,0.8207,0.8337],[0.8713,0.8742,0.8870]),
"70B Few Code":([0.8307,0.8182,0.8534],[0.7980,0.7846,0.8264],[0.8573,0.8494,0.8803]),
},
"Statement of\nNegative Emotion":{
"8B Zero Basic":([0.2467,0.2426,0.3005],[0.1909,0.1788,0.2334],[0.3115,0.3089,0.3909]),
"8B Zero Code":([0.2744,0.2630,0.4211],[0.2109,0.2003,0.3082],[0.3445,0.3298,0.5453]),
"8B Few Basic":([0.2327,0.2403,0.3286],[0.1754,0.1826,0.2552],[0.2945,0.3048,0.4152]),
"8B Few Code":([0.2296,0.2298,0.4030],[0.1765,0.1751,0.3061],[0.2896,0.2936,0.5097]),
"70B Zero Basic":([0.4192,0.2756,0.2869],[0.3190,0.2147,0.2070],[0.5135,0.3433,0.3513]),
"70B Zero Code":([0.4841,0.3276,0.4654],[0.3914,0.2534,0.3811],[0.5895,0.4070,0.5654]),
"70B Few Basic":([0.2690,0.2393,0.2568],[0.2058,0.1836,0.1951],[0.3433,0.3052,0.3320]),
"70B Few Code":([0.3423,0.2549,0.4444],[0.2668,0.1953,0.3532],[0.4261,0.3226,0.5384]),
},
"Statement of\nPositive Emotion":{
"8B Zero Basic":([0.1304,0.2759,0.2222],[0.0506,0.1371,0.1011],[0.2652,0.4615,0.3965]),
"8B Zero Code":([0.2118,0.3265,0.2712],[0.1189,0.1633,0.1442],[0.3452,0.5098,0.4464]),
"8B Few Basic":([0.1235,0.1538,0.2188],[0.0492,0.0558,0.0935],[0.2585,0.2944,0.3824]),
"8B Few Code":([0.1667,0.2917,0.4000],[0.0690,0.1429,0.1667],[0.2989,0.4826,0.6250]),
"70B Zero Basic":([0.1772,0.2182,0.2778],[0.0821,0.0833,0.1176],[0.3150,0.3889,0.5106]),
"70B Zero Code":([0.3000,0.2857,0.3333],[0.1333,0.1332,0.1111],[0.5333,0.5000,0.5833]),
"70B Few Basic":([0.2500,0.3333,0.2162],[0.1190,0.1788,0.0669],[0.4091,0.5119,0.4337]),
"70B Few Code":([0.2424,0.3243,0.2759],[0.0625,0.1316,0.0870],[0.4375,0.5185,0.5150]),
},
"Statement of\nProgress":{
"8B Zero Basic":([0.5393,0.5217,0.5156],[0.4048,0.3881,0.4157],[0.6472,0.6389,0.6329]),
"8B Zero Code":([0.5714,0.5591,0.6139],[0.4474,0.4292,0.4980],[0.6800,0.6737,0.7215]),
"8B Few Basic":([0.5000,0.4932,0.5714],[0.3456,0.3511,0.4287],[0.6389,0.6349,0.6883]),
"8B Few Code":([0.5867,0.5238,0.5684],[0.4536,0.3825,0.4415],[0.7134,0.6431,0.6739]),
"70B Zero Basic":([0.5316,0.5429,0.6154],[0.3886,0.4066,0.4926],[0.6579,0.6765,0.7436]),
"70B Zero Code":([0.5412,0.5412,0.5053],[0.4118,0.4105,0.3803],[0.6517,0.6604,0.6372]),
"70B Few Basic":([0.5278,0.4359,0.5301],[0.3795,0.3134,0.3896],[0.6708,0.5814,0.6520]),
"70B Few Code":([0.4595,0.4474,0.5366],[0.3111,0.3072,0.3984],[0.5961,0.5876,0.6667]),
},
"Statement of\nChallenge":{
"8B Zero Basic":([0.8114,0.6749,0.8119],[0.7709,0.6191,0.7788],[0.8421,0.7140,0.8476]),
"8B Zero Code":([0.7520,0.8413,0.8381],[0.7197,0.8104,0.8046],[0.7861,0.8711,0.8677]),
"8B Few Basic":([0.8060,0.8071,0.8209],[0.7751,0.7733,0.7861],[0.8387,0.8400,0.8552]),
"8B Few Code":([0.7742,0.8178,0.8481],[0.7422,0.7848,0.8135],[0.8086,0.8510,0.8741]),
"70B Zero Basic":([0.8321,0.8244,0.7725],[0.7920,0.7873,0.7210],[0.8643,0.8570,0.8101]),
"70B Zero Code":([0.8331,0.8711,0.8817],[0.8048,0.8435,0.8523],[0.8598,0.8995,0.9103]),
"70B Few Basic":([0.8649,0.8586,0.8767],[0.8352,0.8218,0.8470],[0.8915,0.8851,0.9020]),
"70B Few Code":([0.8480,0.8321,0.8648],[0.8130,0.7987,0.8353],[0.8732,0.8603,0.8922]),
}
}

# Validate that every F1 value falls within its corresponding confidence interval.
for panel, data in panels.items():
    for key, (vals, lows, highs) in data.items():
        for label, v, lo, hi in zip(prompt_labels, vals, lows, highs):
            assert lo <= v <= hi, (panel, key, label, v, lo, hi)

# ---------- Plot ----------
fig = plt.figure(figsize=(16, 10.5))
gs = fig.add_gridspec(
    nrows=5, ncols=2,
    left=0.19, right=0.985, top=0.84, bottom=0.095,
    hspace=0.36, wspace=0.06
)

axes = np.empty((5,2), dtype=object)
for r in range(5):
    for c in range(2):
        axes[r,c] = fig.add_subplot(gs[r,c])

x = np.arange(3)
offsets = {
    ("Zero","Basic"): -0.27,
    ("Zero","Code"): -0.09,
    ("Few","Basic"): 0.09,
    ("Few","Code"): 0.27,
}
marker_lookup = {"Zero": "o", "Few": "s"}
color_lookup = {"Basic": basic_color, "Code": codebook_color}

panel_names = list(panels.keys())
letters = ["A","B","C","D","E"]

for row, panel_name in enumerate(panel_names):
    for col, model in enumerate(["8B", "70B"]):
        ax = axes[row, col]

        # Draw error bars first, behind markers, with no caps.
        for shot in ["Zero", "Few"]:
            for definition in ["Basic", "Code"]:
                key = f"{model} {shot} {definition}"
                vals, lows, highs = panels[panel_name][key]
                xpos = x + offsets[(shot, definition)]
                ax.errorbar(
                    xpos, vals, yerr=err_array(vals, lows, highs),
                    fmt="none",
                    ecolor=color_lookup[definition],
                    elinewidth=1.6,
                    capsize=0,
                    linestyle="None",
                    zorder=2
                )

        # Draw filled markers last, so circles and squares remain visually distinct.
        for shot in ["Zero", "Few"]:
            for definition in ["Basic", "Code"]:
                key = f"{model} {shot} {definition}"
                vals, lows, highs = panels[panel_name][key]
                xpos = x + offsets[(shot, definition)]
                ax.plot(
                    xpos, vals,
                    marker=marker_lookup[shot],
                    linestyle="None",
                    markersize=6.2,
                    markerfacecolor=color_lookup[definition],
                    markeredgecolor="white",
                    markeredgewidth=0.8,
                    color=color_lookup[definition],
                    zorder=3
                )

        ax.set_ylim(0, 1.0)
        ax.set_xlim(-0.70, 2.70)
        ax.set_yticks(np.arange(0, 1.01, 0.25))
        ax.yaxis.grid(True, color=grid_color, linewidth=0.75, linestyle="--", dashes=(2,2), alpha=0.9)
        ax.xaxis.grid(False)
        ax.set_xticks(x)

        if row == 4:
            ax.set_xticklabels(prompt_labels)
            ax.set_xlabel("Prompt modification", labelpad=10, fontweight="semibold")
        else:
            ax.set_xticklabels([])
            ax.tick_params(axis="x", length=0)

        if col == 0:
            ax.set_ylabel("F1 score", fontweight="semibold")
        else:
            ax.set_yticklabels([])
            ax.tick_params(axis="y", length=0)

        for spine in ["top","right"]:
            ax.spines[spine].set_visible(False)
        for spine in ["left","bottom"]:
            ax.spines[spine].set_color(spine_color)
            ax.spines[spine].set_linewidth(0.8)
        ax.tick_params(colors=text_color)

    bbox = axes[row,0].get_position()
    y_mid = (bbox.y0 + bbox.y1) / 2
    fig.text(0.012, bbox.y1 - 0.002, letters[row],
             ha="left", va="top", fontsize=14, fontweight="bold", color=text_color)
    fig.text(0.034, y_mid, panel_name,
             ha="left", va="center", fontsize=12.5, fontweight="semibold", color=text_color, linespacing=1.15)

# Model headers
for col, model_label in enumerate(["Llama-3.1-8B", "Llama-3.1-70B"]):
    bbox = axes[0,col].get_position()
    x_mid = (bbox.x0 + bbox.x1)/2
    fig.text(x_mid, 0.872, model_label, ha="center", va="center",
             fontsize=14.5, fontweight="semibold", color=text_color)
    fig.lines.append(plt.Line2D([bbox.x0, bbox.x1], [0.852, 0.852],
                                transform=fig.transFigure, color=text_color, linewidth=1.35))

# Top legend
legend_y = 0.956
fig.text(0.205, legend_y, "DEFINITION:", ha="right", va="center",
         fontsize=11.5, fontweight="bold", color=text_color)
fig.text(0.49, legend_y, "|", ha="center", va="center",
         fontsize=18, color=text_color)
fig.text(0.555, legend_y, "EXAMPLES:", ha="right", va="center",
         fontsize=11.5, fontweight="bold", color=text_color)

definition_handles = [
    Line2D([0],[0], marker="o", linestyle="None", color=basic_color,
           markerfacecolor=basic_color, markeredgecolor=basic_color,
           markersize=9, label="Basic"),
    Line2D([0],[0], marker="o", linestyle="None", color=codebook_color,
           markerfacecolor=codebook_color, markeredgecolor=codebook_color,
           markersize=9, label="Codebook"),
]
example_handles = [
    Line2D([0],[0], marker="o", linestyle="None", color=text_color,
           markerfacecolor=text_color, markeredgecolor=text_color,
           markersize=8.5, label="Zero-shot"),
    Line2D([0],[0], marker="s", linestyle="None", color=text_color,
           markerfacecolor=text_color, markeredgecolor=text_color,
           markersize=8.5, label="Few-shot"),
]

fig.legend(handles=definition_handles, loc="center", bbox_to_anchor=(0.335, legend_y),
           ncol=2, frameon=False, handletextpad=0.5, columnspacing=1.6, borderaxespad=0)
fig.legend(handles=example_handles, loc="center", bbox_to_anchor=(0.705, legend_y),
           ncol=2, frameon=False, handletextpad=0.5, columnspacing=1.6, borderaxespad=0)

# ---------- Save ----------
outdir = Path("figures")
outdir.mkdir(exist_ok=True)

png = outdir / "Figure2.png"
pdf = outdir / "Figure2.pdf"
svg = outdir / "Figure2.svg"

fig.savefig(png, dpi=600, bbox_inches="tight", facecolor="white")
fig.savefig(pdf, bbox_inches="tight", facecolor="white")
fig.savefig(svg, bbox_inches="tight", facecolor="white")
plt.close(fig)

print(f"Created files:\n{png}\n{pdf}\n{svg}\nFont used: {font_name}")
```
